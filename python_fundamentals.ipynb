## 1. Variables and Data Types

A **variable** is a name that refers to a value. Python is dynamically typed, so you don't declare a type; it is inferred from the value.

| Type | Example | Description |
|------|---------|-------------|
| `int` | `27` | Whole numbers |
| `float` | `9.8` | Decimal numbers |
| `str` | `"name"` | Text |
| `bool` | `True` | `True` / `False` |

Reassigning a variable simply points the name at a new value (the old value is discarded).

```python
print("name")
```

```python
name = "name"
age = 27

# Reassignment: the variables now refer to new values
name = "name2"
age = 88

cgpa = 9.8
is_student = True

print(name, age)
print(type(age))
print(type(name))
print(type(cgpa))
print(type(is_student))
```

## 2. User Input and Type Casting

`input()` always returns a **string**, even if the user types a number. To do maths with it you must convert it.

- **Type casting (explicit):** you convert manually with `int()`, `float()`, `str()`.
- **Type conversion (implicit):** Python converts automatically when it is safe, e.g. `int + float` gives a `float`.

```python
# name = input("enter your name: ")
name = "name"
print(name)
print("hello " + name)
```

```python
# Explicit type casting
# age = input("AGE: ")
age = "27"
print(age, type(age))          # still a string

new_age = int(age) + 1         # str -> int
print(new_age)
print(float(new_age))          # int -> float
```

```python
# Implicit conversion vs explicit casting
print(1 + 2.5)          # int + float -> float (implicit)
print(1 + int(2.4))     # int(2.4) truncates to 2 (explicit)
```

### Example: sum of two numbers

> Avoid naming a variable `sum`, since that shadows Python's built-in `sum()` function. We use `total` instead.

```python
# a = int(input("enter a: "))
# b = int(input("enter b: "))
a, b = 4, 6
total = a + b
print("sum:", total)
```

## 3. Strings

Strings are **immutable**: methods like `upper()` and `replace()` return a *new* string and never change the original.

| Method / operator | Purpose |
|---|---|
| `.upper()` / `.lower()` | Change case |
| `.find(x)` | Index of first match, or `-1` if not found |
| `.replace(old, new)` | Replace all occurrences |
| `x in s` | Membership test (`True`/`False`) |

```python
name = "name"
grade = 'b'   # single or double quotes both work

print(name.upper())
print(name.lower())
print(name)                     # original is unchanged (immutable)

# find: returns the index of the first match
print(name.find("n"))

# replace
print(name.replace("name", "name2"))
print(name.replace("n", "g"))

# membership: `in` is a keyword
print("a" in name)
```

## 4. Operators

### 4.1 Arithmetic and assignment operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` `*` | Add, subtract, multiply | `5*4` | `20` |
| `/` | True division (always `float`) | `5/3` | `1.666...` |
| `//` | Floor division (integer part) | `5//3` | `1` |
| `%` | Remainder (modulus) | `5%3` | `2` |
| `**` | Power | `5**3` | `125` |

Assignment shorthands: `x += 5` is the same as `x = x + 5` (also `-=`, `*=`, `/=`, ...).

```python
print(5 + 5)
print(5 - 2)
print(5 * 4)
print(5 / 3)
print(5 // 3)   # integer part only
print(5 % 3)
print(5 ** 3)

# Assignment operators
x = 1
x += 5          # x = x + 5
print(x)

y = 3
y -= 1
print(y)

z = 2
z *= 2
print(z)
```

### 4.2 Operator precedence (BODMAS)

Order: parentheses `()` first, then `**`, then `*` `/` `//` `%`, then `+` `-` (left to right for equal precedence).

```python
ans = 2 + 5 * 3        # multiplication first -> 17
print(ans)

ans1 = (2 + 5) * 3     # parentheses first -> 21
print(ans1)
```

### 4.3 Comparison operators

These return a boolean: `>`, `<`, `>=`, `<=`, `==` (equal), `!=` (not equal).

> `=` assigns a value, while `==` compares two values.

```python
print(3 > 2)
print(1 > 2)
print(2 < 1)
print(2 <= 2)
```

```python
print(2 == 2)
print(2 != 5)
print(2 != 2)
```

### 4.4 Logical operators

| Operator | True when |
|---|---|
| `and` | both sides are `True` |
| `or` | at least one side is `True` |
| `not` | flips the value |

```python
stt1 = 3 > 5    # False
stt2 = 3 > 2    # True
print(stt1 or stt2)     # True

stt1 = 3 > 5    # False
stt2 = 3 > 12   # False
print(stt1 or stt2)     # False
print(stt1 and stt2)    # False
print(not True)         # False
```

## 5. Conditional Statements

Use `if` / `elif` / `else` to run code based on conditions. Python uses **indentation** (not braces) to define blocks. Conditions are checked top to bottom and only the first true branch runs.

```python
age = 14
if age >= 18:
    print("adult")
    print("you can drive")
elif age < 18:
    print("not adult")
```

Example: grading with multiple conditions. Chained comparison `60 <= marks < 80` is a neat Python shortcut for `marks >= 60 and marks < 80`.

```python
marks = 60
if marks >= 80:
    print("a")
elif marks < 80 and marks >= 60:
    print("b")
else:
    print("c")
```

## 6. Loops

### 6.1 `range()`

`range(start, stop, step)` produces a sequence of integers. `start` defaults to `0`, `step` defaults to `1`, and `stop` is **excluded**.

```python
nums = range(5)
print(nums)          # a range object, not a list
print(list(nums))    # [0, 1, 2, 3, 4]
```

### 6.2 `while` loop

Repeats as long as its condition is `True`. Make sure something inside the loop changes the condition, otherwise it runs forever.

```python
counter = 1
while counter <= 5:
    print("name")
    counter += 1
```

Multiplying a string by an integer repeats it, which makes pattern printing easy.

```python
# Increasing triangle
i = 1
while i <= 5:
    print(i * "*")
    i += 1
```

```python
# Decreasing triangle
i = 5
while i > 0:
    print(i * "*")
    i -= 1
```

### 6.3 `for` loop

Iterates over any sequence (a `range`, list, string, ...).

```python
nums = range(5)
for i in nums:
    print(i)
```

```python
for i in range(1, 5):     # 1 to 4
    print(i)
```

```python
# Even numbers from 1 to 10
for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```

```python
# Same result using the step argument
for i in range(2, 11, 2):
    print(i)
```

### 6.4 `break` and `continue`

- `break` exits the loop immediately.
- `continue` skips the rest of the current iteration and moves to the next.

```python
# break: stops the loop when i reaches 21
for i in range(1, 31):
    if i == 21:
        break
    if i % 3 == 0:
        print(i)
```

```python
# continue: skips 21 but carries on to 30
for i in range(1, 31):
    if i == 21:
        continue
    if i % 3 == 0:
        print(i)
```

## 7. Data Structures: List, Tuple, Set, Dictionary

| Structure | Syntax | Ordered | Mutable | Duplicates |
|---|---|---|---|---|
| List | `[1, 2]` | Yes | Yes | Allowed |
| Tuple | `(1, 2)` | Yes | **No** | Allowed |
| Set | `{1, 2}` | No | Yes | **Not allowed** |
| Dictionary | `{"a": 1}` | Yes (insertion order) | Yes | Keys must be unique |

### 7.1 List

```python
marks = [98, 97, 67, "A", "B", 92.5]   # can mix types
print(marks)
print(len(marks))
print(marks[3])         # indexing starts at 0
print(marks[-1])        # negative index: -1 last, -2 second last
print(marks[0:3])       # slicing: start included, stop excluded
print(marks[:])         # full copy
print(marks[:2])
for score in marks:
    print(score)
```

```python
marks.append(50)        # add to the end
print(marks)
marks.insert(3, 40)     # insert at index 3
print(marks)
print(67 in marks)      # membership test
```

### 7.2 Tuple

Like a list, but **immutable**: it cannot be changed after creation. Useful for fixed data.

```python
marks = (98, 97, 67, "A", "B", 92.5)
print(marks)
print(marks.count(97))   # how many times 97 appears
print(marks.index(67))   # index of first 67
```

### 7.3 Set

An unordered collection of **unique** items; duplicates are removed automatically.

```python
marks = {67, 98, 97, 97, 98}
print(marks)
print(len(marks))
for score in marks:
    print(score)
```

### 7.4 Dictionary

Stores **key: value** pairs. Keys must be unique (assigning to an existing key overwrites its value), while values can repeat.

```python
marks = {"maths": 98, "eng": 78, "chem": 87}
print(marks)
print(marks["maths"])
for key in marks:
    print(key, marks[key])
```

## 8. Functions

A function is a reusable block of code defined with `def`. It can take **parameters** and hand back a result with `return`.

- `print()` only *displays* a value. The function then returns `None`.
- `return` *gives the value back* so it can be stored or reused.

Below, the original version uses `print` inside the function, so `print(add(3, 4))` shows `7` and then `None`.

```python
# Prints inside the function (returns None)
def add(a, b):
    print(a + b)

print(add(3, 4))   # 7 is printed by the function, then None by the outer print
```

```python
# Better: return the value
def add(a, b):
    return a + b

print(add(3, 4))
```

### Example: GST calculator

Adds 18% GST to a price.

```python
def calc_gst(price):
    new_price = price + price * 0.18
    return new_price

print(calc_gst(100))
```
